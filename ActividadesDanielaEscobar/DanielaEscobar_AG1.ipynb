##Devolución de cambio por técnica voraz

```
'''
N es cantidad a buscar
SM son los valores de las monedas
'''
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor
    print(ValorAcumulado)

    if ValorAcumulado == N:
      return SOLUCION


cambio_monedas(15,[25,10,5,1])
```

## N-Reinas por técnica de vueta atrás


```
def escribe(S):
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X " , end="")
      else:
        print(" - ", end="")


def es_prometedora(SOLUCION,etapa):
  #print(SOLUCION)
  #Si la solución tiene dos valores iguales no es valida => Dos reinas en la misma fila
  for i in range(etapa+1):
    #print("El valor " + str(SOLUCION[i]) + " está " +  str(SOLUCION.count(SOLUCION[i])) + " veces")
    if SOLUCION.count(SOLUCION[i]) > 1:
      return False

    #Verifica las diagonales
    for j in range(i+1, etapa +1 ):
      #print("Comprobando diagonal de " + str(i) + " y " + str(j))
      if abs(i-j) == abs(SOLUCION[i]-SOLUCION[j]) : return False
  return True



def reinas(N, solucion=[], etapa=0):
  if len(solucion) == 0:
      solucion=[0 for i in range(N)]

  for i in range(1, N+1):
    solucion[etapa] = i

    if es_prometedora(solucion, etapa):
      if etapa == N-1:
        print(solucion)
        escribe(solucion)
        print()
      else:
        reinas(N, solucion, etapa+1)
    else:
      None

    solucion[etapa] = 0

reinas(8)
```

## Viaje por el rio. Programación dinámica

```
TARIFAS = [
[0,5,4,3,999,999,999],
[999,0,999,2,3,999,11],
[999,999, 0,1,999,4,10],
[999,999,999, 0,5,6,9],
[999,999, 999,999,0,999,4],
[999,999, 999,999,999,0,3],
[999,999,999,999,999,999,0]
]



################################################################
def Precios(TARIFAS):
################################################################
  #Total de Nodos
  N = len(TARIFAS[0])

  #Inicialización de la tabla de precios
  PRECIOS = [ [9999]*N for i in [9999]*N]
  RUTA = [ [""]*N for i in [""]*N]

  for i in range(0,N-1):
    RUTA[i][i] = i             #Para ir de i a i se "pasa por i"
    PRECIOS[i][i] = 0          #Para ir de i a i se se paga 0
    for j in range(i+1, N):
      MIN = TARIFAS[i][j]
      RUTA[i][j] = i

      for k in range(i, j):
        if PRECIOS[i][k] + TARIFAS[k][j] < MIN:
            MIN = min(MIN, PRECIOS[i][k] + TARIFAS[k][j] )
            RUTA[i][j] = k          #Anota que para ir de i a j hay que pasar por k
        PRECIOS[i][j] = MIN

  return PRECIOS,RUTA
################################################################


PRECIOS,RUTA = Precios(TARIFAS)
#print(PRECIOS[0][6])

print("PRECIOS")
for i in range(len(TARIFAS)):
  print(PRECIOS[i])

print("\nRUTA")
for i in range(len(TARIFAS)):
  print(RUTA[i])

#Determinar la ruta con Recursividad
def calcular_ruta(RUTA, desde, hasta):
  if desde == hasta:
    #print("Ir a :" + str(desde))
    return ""
  else:
    return str(calcular_ruta( RUTA, desde, RUTA[desde][hasta])) +  \
                ',' + \
                str(RUTA[desde][hasta] \
              )

print("\nLa ruta es:")
calcular_ruta(RUTA, 0,6)
```

```
#Fibonacci con programación dinámica
def fibonnaci_din(N):
  anterior = 1
  actual = 1
  for _ in range(2,N+1):
    tmp = actual
    actual += anterior
    anterior = tmp
  return actual

print(fibonnaci_din(5))
print(fibonnaci_din(1))
print(fibonnaci_din(0))
print(fibonnaci_din(3))
print(fibonnaci_din(8))
```

Es una mejora de la complejidad con programación dinámica del algoritmo de fibonacci de O(2^n) a lineal O(n)

# Reto propuesto:  Encontrar los dos puntos más cercanos


*   Dado un conjunto de puntos se trata de encontrar los dos puntos más cercanos
*   Guía para aprendizaje:
    * Suponer en 1D, o sea, una lista de números: [3403, 4537, 9089, 9746, 7259, ….
    * Primer intento: Fuerza bruta

    * Segundo intento. Aplicar Divide y Vencerás
    * Calcular la complejidad. ¿Se puede mejorar?
    * Extender el algoritmo a 2D: [(1122, 6175), (135, 4076), (7296, 2741)…
    * Extender el algoritmo a 3D.

## Respuesta
Estoy presuponiendo no ordenar los puntos, sobretodo para que sea uniforme los 1D con los 2D y 3D.

```
def puntos_cercanos_1D_fuerza_bruta(lista):
  minimos = ()
  dis_min = float("inf")
  for i in range(len(lista)):
    for elem in lista[i+1:]:
      if abs(lista[i]-elem) < dis_min:
        minimos = (lista[i],elem)
        dis_min = abs(lista[i]-elem)
  return minimos, dis_min


minimos, dis_min=puntos_cercanos_1D_fuerza_bruta([3403,4537,9089,9746,7259])
print(f'la distancia minima es entre {minimos[0]} y {minimos[1]}: {dis_min}')
```

Aunque en el segundo bucle no se vuelven a recorrer los elementos anteriores, sigue siendo de complejidad cuadrática al tener que recorrer la lista de forma anidada, comparando todos los elementos con todos los demás.

```
def puntos_cercanos_1d_divide_venceras(lista):
  if len(lista) == 1:
    return (lista[0],lista[0]), float("inf")

  if len(lista) == 2:
    return (lista[0],lista[1]),abs(lista[0]-lista[1])

  mitad = len(lista) // 2
  izquierda = lista[:mitad]
  derecha = lista[mitad:]

  minimos_iz, min_iz = puntos_cercanos_1d_divide_venceras(izquierda)
  minimos_der, min_der = puntos_cercanos_1d_divide_venceras(derecha)

  minimos = ()
  dis_min = float("inf")

  if min_iz < min_der:
    minimos = minimos_iz
    dis_min = min_iz
  else:
    minimos = minimos_der
    dis_min = min_der

  for iz in izquierda:
    for der in derecha:
      if abs(iz-der) < dis_min:
        minimos = (iz,der)
        dis_min = abs(iz-der)

  return minimos, dis_min

minimos, dis_min=puntos_cercanos_1d_divide_venceras([3403,4537,9089,9746,7259])
print(f'la distancia minima es entre {minimos[0]} y {minimos[1]}: {dis_min}')
```

La complejidad de mantiene cuadrática al tener que recombinar los puntos, por los casos en que haya un punto de una mitad con la otra sean el de menor distancia.

```
import math

# Fuerza bruta con dos dimensiones
def distancia_2d(p1, p2):
    return math.sqrt(
        (p1[0] - p2[0]) ** 2 +
        (p1[1] - p2[1]) ** 2
    )

def puntos_cercanos_2D_fuerza_bruta(lista):
  minimos = ()
  dis_min = float("inf")
  for i in range(len(lista)):
    for elem in lista[i+1:]:
      if distancia_2d(lista[i],elem) < dis_min:
        minimos = (lista[i],elem)
        dis_min = distancia_2d(lista[i],elem)
  return minimos, dis_min



minimos, dis_min=puntos_cercanos_2D_fuerza_bruta([(1122, 6175), (135, 4076), (7296, 2741)])
print(f'la distancia minima es entre {minimos[0]} y {minimos[1]}: {dis_min}')
```

```
# Fuerza bruta de 3 dimensiones
def distancia_3d(p1, p2):
    return math.sqrt(
        (p1[0] - p2[0]) ** 2 +
        (p1[1] - p2[1]) ** 2 +
        (p1[2] - p2[2]) ** 2
    )

def puntos_cercanos_3D_fuerza_bruta(lista):
  minimos = ()
  dis_min = float("inf")
  for i in range(len(lista)):
    for elem in lista[i+1:]:
      if distancia_3d(lista[i],elem) < dis_min:
        minimos = (lista[i],elem)
        dis_min = distancia_3d(lista[i],elem)
  return minimos, dis_min



minimos, dis_min=puntos_cercanos_3D_fuerza_bruta([(1122, 6175,0), (135, 4076,0), (7296, 2741,0)])
print(f'la distancia minima es entre {minimos[0]} y {minimos[1]}: {dis_min}')
```
